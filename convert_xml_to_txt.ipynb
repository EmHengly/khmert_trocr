```python
import os
import xml.etree.ElementTree as ET

# Define the input XML directory path
input_xml_directory = 'cropped_dataset/xml'

# Define the output text file path
output_text_file = 'data_test.txt'

# Initialize an empty list to store the output lines
output_lines = []

# Iterate over all files in the input directory
for filename in os.listdir(input_xml_directory):
    if filename.endswith('.xml'):
        # Construct the full file path
        file_path = os.path.join(input_xml_directory, filename)
        
        # Parse the XML file
        tree = ET.parse(file_path)
        root = tree.getroot()

        # Extract the image name and text
        image_name = root.find('image').text
        text = root.find('word').find('text').text

        # Create the text content in the specified format
        output_content = f"{image_name}  {text}"
        
        # Append the content to the output lines list
        output_lines.append(output_content)

# Save all the output lines to the text file
with open(output_text_file, 'w', encoding='utf-8') as file:
    for line in output_lines:
        file.write(line + '\n')

print(f"Metadata from all XML files has been saved to {output_text_file}")
```

```python
from PIL import Image
import os

def convert_and_rename_png_to_jpg(source_folder, destination_folder):
    # Create destination folder if it doesn't exist
    if not os.path.exists(destination_folder):
        os.makedirs(destination_folder)
    
    # Get list of PNG files in the source folder
    png_files = [f for f in os.listdir(source_folder) if f.endswith('.png')]

    # Counter for renaming files
    counter = 0

    # Convert and rename PNG files to JPG
    for png_file in png_files:
        # Open PNG image
        with Image.open(os.path.join(source_folder, png_file)) as img:
            # Convert to JPG
            jpg_file = f"{counter:06d}.jpg"
            img = img.convert('RGB')
            # Save as JPG in the destination folder
            img.save(os.path.join(destination_folder, jpg_file))
            # Increment counter
            counter += 1

source_folder = "dataset_png/data_train/"
destination_folder = "dataset/data_train"

convert_and_rename_png_to_jpg(source_folder, destination_folder)
```

```python
import os

def rename_jpg_files_in_text_file(source_folder, file_list_path, start_number):
    # Read the file list
    with open(file_list_path, 'r', encoding='utf-8') as file:
        lines = file.readlines()

    # Counter for renaming files
    counter = start_number
    new_lines = []

    # Modify the filenames in the text file
    for i in range(len(lines)):
        # Split each line into filename and description
        parts = lines[i].split(' ', 1)
        if len(parts) > 1 and parts[0].endswith('.jpg'):
            # Check if the file exists
            old_filename = parts[0]
            old_filepath = os.path.join(source_folder, old_filename)
            if os.path.exists(old_filepath):
                # Create new filename
                new_filename = f"{counter:06d}.jpg"
                new_filepath = os.path.join(source_folder, new_filename)
                # Rename the file
                os.rename(old_filepath, new_filepath)
                # Update the line with new filename
                new_lines.append(new_filename + ' ' + parts[1])
                # Increment counter
                counter += 1
            else:
                # If file doesn't exist, keep the line unchanged
                new_lines.append(lines[i])
        else:
            # If line format is incorrect, keep the line unchanged
            new_lines.append(lines[i])

    # Write the modified lines back to the text file
    with open(file_list_path, 'w', encoding='utf-8') as file:
        file.writelines(new_lines)

source_folder = "dataset/data_test"
file_list_path = "dataset/data_test.txt"
start_number = 22567

rename_jpg_files_in_text_file(source_folder, file_list_path, start_number)
```

```python
import os

def get_jpg_filenames_from_text(file_list_path):
    with open(file_list_path, 'r', encoding='utf-8') as file:
        lines = file.readlines()
    jpg_filenames = [line.split(' ')[0] for line in lines if line.strip().endswith('.jpg')]
    return jpg_filenames

def find_missing_jpg_files(directory, start_number, max_number):
    missing_files = []
    for number in range(start_number, max_number + 1):
        filename = f"{number:06d}.jpg"
        if not os.path.exists(os.path.join(directory, filename)):
            missing_files.append(filename)
    return missing_files

def main():
    source_folder = "dataset/data_test"
    file_list_path = "dataset/data_test.txt"

    jpg_filenames = get_jpg_filenames_from_text(file_list_path)
    start_number = int(jpg_filenames[0].split('.')[0]) if jpg_filenames else 1
    max_number = int(jpg_filenames[-1].split('.')[0]) if jpg_filenames else 0

    missing_files = find_missing_jpg_files(source_folder, start_number, max_number)

    if missing_files:
        print("Missing JPG files:")
        for filename in missing_files:
            print(filename)
    else:
        print("No JPG files are missing.")

if __name__ == "__main__":
    main()
```
