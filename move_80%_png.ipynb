```python
import os
import random
import shutil

def move_files(source_folder, target_folder, percentage=80):
    # Ensure target folder exists
    if not os.path.exists(target_folder):
        os.makedirs(target_folder)

    # List all PNG files in the source folder
    files = [f for f in os.listdir(source_folder) if f.endswith('.png')]
    
    # Calculate the number of files to move
    num_files_to_move = int(len(files) * (percentage / 100))
    
    # Randomly select files to move
    files_to_move = random.sample(files, num_files_to_move)
    
    # Move the files
    for file in files_to_move:
        shutil.move(os.path.join(source_folder, file), os.path.join(target_folder, file))
    
    print(f"Moved {len(files_to_move)} files from {source_folder} to {target_folder}")

# Example usage
source_folder = 'cropped_dataset/png/'
target_folder = 'data_train'
move_files(source_folder, target_folder, 80)
```

```python
import os
import random
import shutil

def move_files_with_pairs(source_png_folder, source_xml_folder, target_png_folder, target_xml_folder, percentage=80):
    # Ensure target folders exist
    if not os.path.exists(target_png_folder):
        os.makedirs(target_png_folder)
    if not os.path.exists(target_xml_folder):
        os.makedirs(target_xml_folder)

    # List all PNG and XML files in the source folders
    png_files = {f[:-4]: f for f in os.listdir(source_png_folder) if f.endswith('.png')}
    xml_files = {f[:-4]: f for f in os.listdir(source_xml_folder) if f.endswith('.xml')}
    
    # Find the common base names between PNG and XML files
    common_files = set(png_files.keys()) & set(xml_files.keys())
    
    # Calculate the number of pairs to move
    num_files_to_move = int(len(common_files) * (percentage / 100))
    
    # Randomly select pairs to move
    files_to_move = random.sample(common_files, num_files_to_move)
    
    # Move the files
    for base_name in files_to_move:
        shutil.move(os.path.join(source_png_folder, png_files[base_name]), os.path.join(target_png_folder, png_files[base_name]))
        shutil.move(os.path.join(source_xml_folder, xml_files[base_name]), os.path.join(target_xml_folder, xml_files[base_name]))
    
    print(f"Moved {len(files_to_move)} pairs of files from {source_png_folder} and {source_xml_folder} to {target_png_folder} and {target_xml_folder}")

# Example usage
source_png_folder = 'cropped_dataset/png/'
source_xml_folder = 'cropped_dataset/xml/'
target_png_folder = 'data_train'
target_xml_folder = 'data_train_xml/'
move_files_with_pairs(source_png_folder, source_xml_folder, target_png_folder, target_xml_folder, 80)
```

```python
import os

def check_file_pairs(png_folder, xml_folder):
    # List all PNG and XML files in the target folders
    png_files = {f[:-4] for f in os.listdir(png_folder) if f.endswith('.png')}
    xml_files = {f[:-4] for f in os.listdir(xml_folder) if f.endswith('.xml')}
    
    # Check for missing pairs
    missing_png_files = xml_files - png_files
    missing_xml_files = png_files - xml_files
    
    # Print results
    if not missing_png_files and not missing_xml_files:
        print("All files are correctly paired.")
    else:
        if missing_png_files:
            print("Missing PNG files for the following XML files:")
            for base_name in missing_png_files:
                print(f"  {base_name}.xml")
        if missing_xml_files:
            print("Missing XML files for the following PNG files:")
            for base_name in missing_xml_files:
                print(f"  {base_name}.png")

# Example usage
target_png_folder = 'data_train/'
target_xml_folder = 'data_train_xml/'
check_file_pairs(target_png_folder, target_xml_folder)
```

```python
import os

def check_and_print_file_pairs(png_folder, xml_folder):
    # List all PNG and XML files in the target folders
    png_files = {f[:-4]: f for f in os.listdir(png_folder) if f.endswith('.ng')}
    xml_files = {f[:-4]: f for f in os.listdir(xml_folder) if f.endswith('.xml')}
    
    # Find the common base names between PNG and XML files
    common_files = set(png_files.keys()) & set(xml_files.keys())
    
    # Check for missing pairs
    missing_png_files = set(xml_files.keys()) - common_files
    missing_xml_files = set(png_files.keys()) - common_files
    
    # Print each pair and check for missing pairs
    if common_files:
        print("Paired files:")
        for base_name in sorted(common_files):
            print(f"  {png_files[base_name]} <-> {xml_files[base_name]}")
    
    if not missing_png_files and not missing_xml_files:
        print("\nAll files are correctly paired.")
    else:
        if missing_png_files:
            print("\nMissing PNG files for the following XML files:")
            for base_name in sorted(missing_png_files):
                print(f"  {xml_files[base_name]}")
        if missing_xml_files:
            print("\nMissing XML files for the following PNG files:")
            for base_name in sorted(missing_xml_files):
                print(f"  {png_files[base_name]}")

# Example usage
target_png_folder = 'data_train/'
target_xml_folder = 'data_train_xml/'
check_and_print_file_pairs(target_png_folder, target_xml_folder)
```
